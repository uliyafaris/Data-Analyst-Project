```python
# Data manipulation and analysis using the DataFrame structure
import pandas as pd

# Python libraries for data visualization (creating graphs, diagrams, and statistical plots)
import seaborn as sns
import matplotlib.pyplot as plt

#link datasets: https://www.kaggle.com/datasets/srisyra02/ecommerce-customer-behavior-dataset?resource=download
```

```python
# Retrieve and display data sets
df = pd.read_csv(r"D:\Data Analyst\Python\Ecommerce Customer Behavior\eCommerce-dataset.csv")
df
```

```python
# To display all rows of a dataframe
# pd.set_option('display.max.rows', 50000)
pd.set_option('display.max.columns', 25)
```

```python
df.info()
```

```python
# Displays if there is data that is null/empty
df.isnull().sum()
```

```python
# Displaying unique data
df.nunique()
```

```python
# Remove data duplicated
df = df.drop_duplicates()
```

```python
# Removing unused/unnecessary columns
df = df.drop(columns = "Churned")
df
```

```python
# Sorting the data to show the 5 lowest values for 'Age' using the head() function
df.sort_values(by='Age').head()
```

```python
# Converting a float data type to an integer
df['Age'] = df['Age'].astype('Int64')
df
```

```python
df.sort_values(by='Age', ascending=False).head()
```

```python
# Removing and then replacing 'Age' values ​​> 100 with the median value, as 'Age' entries of 150 and 100 were previously identified.

import numpy as np
# 1. Change implausible ages (> 100 or < 0) to NaN
df.loc[(df['Age'] > 100) | (df['Age'] < 0), 'Age'] = np.nan

# 2. Calculate the median of the cleaned age data (excluding the value 200)
median_age = df['Age'].median()

# 3. Fill the NaN values ​​with that median
df['Age'] = df['Age'].fillna(median_age)
```

```python
# View rows with age above 100 or below 0
df[(df['Age'] > 100) | (df['Age'] < 0)]
```

```python
# Results after the age data has been cleaned
df.sort_values(by='Age', ascending=False).head()
```

```python
# Displaying the bottom 10 records
df.tail(10)
```

```python
print(plt.style.available)
plt.style.use('Solarize_Light2')
```

```python
# Setting theme
sns.set_theme(style="whitegrid")

# Defining numerical interval limits to divide age data into several ranges
bins = [0, 25, 40, 60, 100]

# Specify a label name for each of the bin ranges defined above. 
# The number of labels must always be one less than the number of bins (4 labels for 5 boundaries).
labels = ["<=25", "26-40", "41-60", "60+"]

# Convert the numerical values ​​in the `df["Age"]` column into age group categories based on 
# specified bins and labels, and store them in a new column named "Age_Group" within the DataFrame `df`.
df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels)

# Create a new chart canvas with a width of 8 inches and a height of 5 inches so that the text and the chart bars do not overlap
plt.figure(figsize=(8, 5))
# Creating a bar chart that automatically calculates the number of rows/users in each category on the X-axis
sns.countplot(data=df, x="Age_Group", hue="Gender")
# Menambahkan judul di bagian atas grafik.
plt.title("Number of Users by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("Total")
plt.show()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Gender", palette="pastel")
plt.title("Number of Data Entries by Gender")
plt.xlabel("Gender")
plt.ylabel("Total")
plt.show()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 10))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix of Customer Numerical Features")
plt.show()
```
